# Palindromes
Un notebook garde l'état entre les cellules.

```python
def est_palindrome(s):
    """Vrai si s se lit dans les deux sens, espaces et casse ignorées."""
    s = s.lower().replace(" ", "")
    return s == s[::-1]
```

```python
phrases = ["kayak", "radar", "python", "Engage le jeu que je le gagne"]
[p for p in phrases if est_palindrome(p)]
```

```python
est_palindrome("Ésope reste ici et se repose")
```
